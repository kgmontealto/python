### Part 1: Building the base Anagram Checker

Given two valid strings, check to see if they are anagrams of each other. If it is, return `True`, else `False`. For this part, we can assume that uppercase letters are the same as if it was a lowercase character.

Examples of anagrams:
* Silent and Listen
* Night and Thing

Example outputs:
```python
anagram_checker("Silent", "listen") # True
anagram_checker("Silent", "Night") # False
anagram_checker("night", "Thing") # True
```

````python
# For testing purposes, we will write our code in the function
def anagram_checker(word_a, word_b):
  # Your code here

    """
    
    This function checks whether two words or phrases 
    are anagrams of each other without case sensitivity

    Examples:

    anagram_checker("save", "Vase")
    -> False: save and Vase are anagrams of each other
    
    anagram_checker("Silent", "Night")
    -> True: Silent and Night are not anagrams of each other

    """

    # Save original input for the result message.
    original_word_a = word_a
    original_word_b = word_b

    # Prepare inputs for comparison
    word_a = word_a.lower().replace(" ","")
    word_b = word_b.lower().replace(" ","")

    # Compare the sorted characters
    if (sorted(word_a) == sorted(word_b)) == True:
        return f"True: {original_word_a} and {original_word_b} are anagrams of each other" 
    else:
         return f"False: {original_word_a} and {original_word_b} are not anagrams of each other"

# Run your code to check using the words below:
anagram_checker("Silent", "listen")
````

### Part 2: Expanding the functionality of the Anagram Checker

Using your existing and functional anagram checker, let's add a boolean option called `is_case_sensitive`, which will return `True` or `False` based on if the two compared words are anagrams and if we are checking for case sensitivity.

````python
def anagram_checker(word_a, word_b, is_case_sensitive):
  # Modify your existing code here

  """
      This function checks whether two words or phrases are anagrams 
      of each other, with the option to choose for case sensitivity
  
      Examples:
      anagram_checker("save", "vase", "yes")
      -> True: save and vase are anagrams of each other
      
      anagram_checker("save", "Vase", "yes")
      -> False: save and Vase are not anagrams of each other
      
      anagram_checker("save", "Vase", "no")
      -> True: save and Vase are anagrams of each other
  
      """
  
  # Save original input for the result message.
  original_word_a = word_a
  original_word_b = word_b
      
  # Standardize the option
  is_case_sensitive = is_case_sensitive.lower()
  
  # Valid case-sensitivity options.
  case_sensitivity_options = {"yes":True
                              ,"y":True
                              ,"true":True
                              ,"t":True
                              ,"no":False
                              ,"n":False
                              ,"false":False
                              ,"f":False}
  
  
  # Check for invalid option. 
  if is_case_sensitive not in case_sensitivity_options:
    return print("You entered a wrong case sensitivity option.\n" 
                 "You can only use (yes, y, no, n, true, t, false or f)")
  
  # Prepare inputs for comparison
  if case_sensitivity_options[is_case_sensitive] == True:
    word_a = word_a.replace(" ","")
    word_b = word_b.replace(" ","")
  else:
    word_a = word_a.lower().replace(" ","")
    word_b = word_b.lower().replace(" ","")
  
  # Compare the sorted characters
  if (sorted(word_a) == sorted(word_b)) == True:
    return f"True: {original_word_a} and {original_word_b} are anagrams of each other" 
  else:
     return f"False: {original_word_a} and {original_word_b} are not anagrams of each other"

# Run your code to check using the words below:
anagram_checker("Silent", "listen", False) # True
````

````python
anagram_checker("Silent", "listen", True) # False
````

````python
anagram_checker("Silent", "Listen", True) # False
````
